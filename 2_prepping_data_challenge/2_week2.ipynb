Join and put the columns in order

```python
%%sql

SELECT
    "Transaction ID",
    'GB' AS "Country",
    "Check Digits",
    "Swift Code",
    REPLACE(t."Sort Code", '-', '') AS "Sort Code",
    "Account Number"
FROM read_csv_auto('dataset/Transactions.csv') AS t
JOIN read_csv_auto('dataset/Swift Codes.csv') AS s
    ON t.Bank = s.Bank;
```

Count number of rows

```python
%%sql
SELECT COUNT(*) AS rows
FROM
(SELECT
    "Transaction ID",
    'GB' AS "Country",
    "Check Digits",
    "Swift Code",
    REPLACE(t."Sort Code", '-', '') AS "Sort Code",
    "Account Number"
FROM read_csv_auto('dataset/Transactions.csv') AS t
JOIN read_csv_auto('dataset/Swift Codes.csv') AS s
    ON t.Bank = s.Bank
);
```

Merge columns and create specified output

```python
%%sql

SELECT
    "Transaction ID",
    CONCAT(
    'GB',
    "Check Digits",
    "Swift Code",
    REPLACE(t."Sort Code", '-', ''),
    "Account Number") AS IBAN
FROM read_csv_auto('dataset/Transactions.csv') AS t
JOIN read_csv_auto('dataset/Swift Codes.csv') AS s
    ON t.Bank = s.Bank;
```
