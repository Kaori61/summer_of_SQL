# 2023: Week 1 The Data Source Bank
Using SQL to clean and transform banking transaction data.
<br>
The challenge requirements and dataset can be downloaded from [Preppin' Data Challenge](https://preppindata.blogspot.com/2023/01/2023-week-1-data-source-bank.html)

```python
# Set up SQL
%load_ext sql
%sql duckdb:///project.duckdb
```

```python
%%sql

SELECT *
FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
LIMIT 10;
```

### Output 1

```python
%%sql

SELECT
    SPLIT_PART("Transaction Code", '-', 1) AS Bank,
    SUM(Value) AS Value
FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
GROUP BY Bank;
```

### Output 2

```python
%%sql
SELECT
    SPLIT_PART("Transaction Code", '-', 1) AS Bank,
    CASE
        WHEN "Online or In-Person" = 1 then 'Online'
        WHEN "Online or In-Person" = 2 then 'In-Person'
    END AS "Online or In-Person",
    DAYNAME("Transaction Date") AS "Transation Date",
    SUM(Value) AS Value
FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
GROUP BY 1,2,3;
```

```python
%%sql
# Count the number of rows in Output 2

SELECT
    COUNT(*) as Total_rows
FROM (
    SELECT
    SPLIT_PART("Transaction Code", '-', 1) AS Bank,
    CASE
        WHEN "Online or In-Person" = 1 then 'Online'
        WHEN "Online or In-Person" = 2 then 'In-Person'
    END AS "Online or In-Person",
    DAYNAME("Transaction Date") AS "Transation Date",
    SUM(Value) AS Value
    FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
    GROUP BY 1,2,3
);
```

### Output 3

```python
%%sql
# check the distinct values of "Customer Code" column

SELECT
    DISTINCT "Customer Code"
FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv');
```

```python
%%sql

SELECT
    SPLIT_PART("Transaction Code", '-', 1) AS Bank,
    "Customer Code",
    SUM(Value) AS Value
FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
GROUP BY 1,2;
```

```python
%%sql
# Show the total number of rows in the output 3

SELECT COUNT(*) as total_rows
FROM(
    SELECT
        SPLIT_PART("Transaction Code", '-', 1) AS Bank,
        "Customer Code",
        SUM(Value) AS Value
    FROM read_csv_auto('dataset/PD_2023_Wk_1_Input.csv')
    GROUP BY 1,2
);
```
